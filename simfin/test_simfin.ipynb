# Exemple de l'utilisation de SimFin

```python
# Importer les librairies nécessaires
import warnings
import pandas as pd 
import numpy as np
from matplotlib import pyplot as plt
warnings.filterwarnings("ignore")
```

```python
# Importer les librairies de simfin
import simfin
```

```python
# Année de début et de fin de la simulation
yr_deb = 2021
yr_fin = 2050
```

```python
# Ajustement de l'index des années. Cela devrait être ajuster dans une prochaine version
yr_fin +=1
```

```python
# Création de l'objet simulateur
x = simfin.simulator(yr_deb,yr_fin)
# Lancer la simulation
x.simulate(yr_fin-yr_deb)
```

```python
# # Les revenus du gouvernement du Québec (en millions de $ courants)
x.summary_fr
```

```python
# Les revenus du gouvernement du Québec (en millions de $ courants)
table = pd.DataFrame()
outcomes = ['Impôt des particuliers','Impôt des sociétés','Taxes à la consommation','Revenus divers','Droits et permis','Cotisations au FSS','Entreprises du gouvernement','Impôt foncier scolaire','Revenus autonomes','Transferts fédéraux COVID','Transferts fédéraux hors COVID','Transferts fédéraux','Total des revenus']
for i in range(2021,yr_fin):
    for o in outcomes:
        table.loc[o,i] = round(x.summary_fr.loc[o,i],1)
table
```

```python
# Les dépenses du gouvernement du Québec (en millions de $ courants)
table = pd.DataFrame()
outcomes = ['Santé et services sociaux','Éducation et culture','Soutien aux personnes et aux familles','Économie et environnement','Gouverne et justice','Dépenses COVID','Dépenses des missions','Service de la dette','Total des dépenses']
for i in range(2021,yr_fin):
    for o in outcomes:
        table.loc[o,i] = round(x.summary_fr.loc[o,i],1)
table
```

```python
# Le surplus (déficit) budgétaire et le solde budgétaire du gouvernement du Québec (en millions de $ courants)
table = pd.DataFrame()
outcomes = ['Total des revenus','Total des dépenses','Surplus annuel','Contributions FDG','Solde budgétaire']
for i in range(2021,yr_fin):
    for o in outcomes:
        table.loc[o,i] = round(x.summary_fr.loc[o,i],1)
table
```

```python
# L'évolution de la dette brute, des dépenses totales et des revenus totaux en % du PIB
outcomes = ['gross debt','total spending','total revenue']
label = ['Dette brute','Dépenses totales','Revenus totaux']
labels = dict(zip(outcomes,label))
plt.figure()
for o in outcomes:
    plt.plot(x.summary.loc[o,:]/x.summary.loc['gdp',:]*100, label=labels[o])
plt.xlabel('Année')
plt.ylabel('En % du PIB')
plt.legend()
plt.title(str(label[2]) + ', ' + str(label[1]) + ' et ' + str(label[0]) + ' :\n Evolution entre 2021 et ' + str(yr_fin) + ' (en % du PIB)')
plt.show()
```

```python
# Le passage de la dette directe consolidée à la dette brute (en millions de $ courants)
table = pd.DataFrame()
for i in range(2021,yr_fin):
    table.loc['Dette directe consolidée',i] = round(x.summary_fr.loc['Dette directe consolidée',i],1)
    table.loc['- Solde FDG',i] = round(x.summary_fr.loc['Solde FDG',i],1)
    table.loc['+ Passif net des régimes de retraite',i] = round(x.summary_fr.loc['Passif net des régimes de retraite',i],1)
    table.loc['= Dette brute',i] = round(x.summary_fr.loc['Dette brute',i],1)
table
```

```python
# La dynamique de la dette directe consolidée (en millions de $ courants)
table = pd.DataFrame()
for i in range(2022,yr_fin):
    table.loc['Dette directe consolidée en début de période',i] = round(x.summary_fr.loc['Dette directe consolidée',i-1],1)
    table.loc['+ Flux placements',i] = round(x.summary_fr.loc['Flux placements',i],1)
    table.loc['+ Flux immobilisations',i] = round(x.summary_fr.loc['Flux immobilisations',i],1)
    table.loc['+ Flux autres',i] = round(x.summary_fr.loc['Flux autres',i],1)
    table.loc['- Solde budgétaire',i] = round(x.summary_fr.loc['Solde budgétaire',i],1)
    table.loc['- Retraits FDG',i] = round(x.summary_fr.loc['Retraits FDG',i],1)
    table.loc['- ∆ Passif net des régimes de retraite',i] = round(x.summary_fr.loc['Passif net des régimes de retraite',i],1)-round(x.summary_fr.loc['Passif net des régimes de retraite',i-1],1)
    table.loc['= Dette directe consolidée en fin de période',i] = round(x.summary_fr.loc['Dette directe consolidée',i],1)
table
```

```python
# La dynamique de la Dette brute (en millions de $ courants)
table = pd.DataFrame()
for i in range(2022,yr_fin):
    table.loc['Dette brute en début de période',i] = round(x.summary_fr.loc['Dette brute',i-1],1)
    table.loc['+ Flux placements',i] = round(x.summary_fr.loc['Flux placements',i],1)
    table.loc['+ Flux immobilisations',i] = round(x.summary_fr.loc['Flux immobilisations',i],1)
    table.loc['+ Flux autres',i] = round(x.summary_fr.loc['Flux autres',i],1)
    table.loc['- Solde budgétaire',i] = round(x.summary_fr.loc['Solde budgétaire',i],1)
    table.loc['- Contributions FDG',i] = round(x.summary_fr.loc['Contributions FDG',i],1)
    table.loc['= Dette brute en fin de période',i] = round(x.summary_fr.loc['Dette brute',i],1)
table
```

## Scénarios alternatifs

```python
# On modifie la productivité totale des facteurs "macro.gr_A" et le taux d'inflation "macro.infl"
y = simfin.simulator(yr_deb,yr_fin)
y.macro.gr_A = 0.007#dans le scénario : 0.02381	en 2021, 0.02260	en 2022, 0.00905 en 2023 et 0.00586 par la suite, ce qui permet de calibrer le taux de croissance réel du PIB en 2021, 2022 et 2023.
y.macro.infl = 0.022#2% dans le sénario de référence (par défaut dans le modèle).
y.simulate(yr_fin-yr_deb)
```

```python
# Graphique de la dette brute du gouvernement du Québec en % du PIB
plt.figure()
plt.plot(x.summary.loc['gross debt',:]/x.summary.loc['gdp',:]*100, label='Scénario de référence')
plt.plot(y.summary.loc['gross debt',:]/y.summary.loc['gdp',:]*100, label='Scénario alternatif')
plt.xlabel('Année')
plt.ylabel('En % du PIB')
plt.title('Dette brute du gouvernement du Québec\nen pourcentage du PIB entre 2021 et ' + str(yr_fin))
plt.legend()
plt.show()
```

Exemple : par défaut les transferts fédéraux en santé croissent au rythme du PIB potentiel nominal. Dans notre exemple, on ajoute à cette hypothèse le taux de croissance de la population en activant le "Switch" ipop de la fonction "health_transfer".

```python
z = simfin.simulator(yr_deb,yr_fin)
z.transfers.health_transfer.ipop=True
z.simulate(yr_fin-yr_deb)
```

```python
# Graphique de la dette brute du gouvernement du Québec en % du PIB
plt.figure()
plt.plot(x.summary.loc['gross debt',:]/x.summary.loc['gdp',:]*100, label='Scénario de référence')
plt.plot(z.summary.loc['gross debt',:]/z.summary.loc['gdp',:]*100, label='Scénario alternatif')
plt.xlabel('Année')
plt.ylabel('En % du PIB')
plt.title('Dette du gouvernement du Québec en pourcentage du PIB\nentre 2021 et ' + str(yr_fin))
plt.legend()
plt.show()
```
